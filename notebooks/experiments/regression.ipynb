```python
import sys
import polars as pl
import plotly.express as px
sys.path.insert(0, '..')
from fs_thesis import sql, show
```

```python
import numpy as np
np.random.seed(69) #This is good practice for reproducibiity
```

```python
# Get patient demographics and admissions (without weight/height for now)
df_lr = sql("""
    SELECT 
        p.subject_id,
        a.hadm_id,
        p.gender,
        p.anchor_age,
        a.race,
        a.insurance,
        a.marital_status,
        a.language
    FROM hosp.patients p
    JOIN hosp.admissions a ON p.subject_id = a.subject_id
""")

# Get all diagnoses
df_diagnoses = sql("SELECT * FROM hosp.diagnoses_icd")

# Create target variable
has_A0 = (
    df_diagnoses
    .with_columns(pl.col("icd_code").str.starts_with("A0").alias("has_A0"))
    .group_by("subject_id")
    .agg(pl.col("has_A0").max().alias("has_A0"))
)

# Join and create final dataset
df_lr = (
    df_lr
    .join(has_A0, on="subject_id", how="left")
    .with_columns(pl.col("has_A0").fill_null(False).cast(pl.Int8))
    .fill_null("UNKNOWN")
)

show(df_lr, limit=True)
```

```python


fig = px.scatter(df_lr, x="anchor_age", y="insurance", color='has_A0')
fig.show()

```
